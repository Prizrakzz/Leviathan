````python
_PAYLOAD_B64 = ""

import base64, json, zlib
import numpy as np
import pandas as pd
from IPython import get_ipython

_IPYTHON = get_ipython()
if _IPYTHON is not None:
    _IPYTHON.run_line_magic("matplotlib", "inline")

import matplotlib.pyplot as plt
from IPython.display import Markdown, display

PAYLOAD = json.loads(zlib.decompress(base64.b64decode(_PAYLOAD_B64)).decode("utf-8"))
SUMMARY = PAYLOAD["summary"]
OVERLAY = PAYLOAD["overlay"]
CONTRACT = PAYLOAD["contract"]
PROVENANCE = PAYLOAD["provenance"]
FRAME_URI = PAYLOAD.get("frame_uri")
MANIFEST_URI = PAYLOAD.get("manifest_uri")
EXACTNESS = SUMMARY["profile"]["analysis_exactness"]
TABLE_NAME = SUMMARY["table_name"]

PREVIEW_ROWS = 20
MAX_NUMERIC_CHARTS = 12
MAX_CATEGORICAL_CHARTS = 8
CHART_MAX_POINTS = 200000

pd.set_option("display.max_rows", PREVIEW_ROWS)
pd.set_option("display.max_columns", 30)
pd.set_option("display.max_colwidth", 100)
plt.rcParams.update({
    "figure.dpi": 120,
    "savefig.dpi": 120,
    "figure.facecolor": "white",
    "axes.facecolor": "white",
    "axes.grid": True,
    "grid.alpha": 0.22,
    "font.size": 9,
})

def metric(section, name, default=None):
    return (SUMMARY.get("profile", {}).get("sections", {}).get(section, {})
            .get("metrics", {}).get(name, {"value": default, "exactness": EXACTNESS}))

def metric_value(section, name, default=None):
    return metric(section, name, default).get("value", default)

def bounded(value, chars=900):
    if isinstance(value, (dict, list, tuple)):
        text = json.dumps(value, sort_keys=True, default=str)
    else:
        text = str(value)
    return text if len(text) <= chars else text[:chars] + " … [bounded display]"

def show_metrics(section, limit=30):
    rows = []
    metrics = (SUMMARY.get("profile", {}).get("sections", {}).get(section, {})
               .get("metrics", {}))
    for name, item in sorted(metrics.items()):
        rows.append({
            "metric": name,
            "value": bounded(item.get("value")),
            "unit": item.get("unit") or "—",
            "n": item.get("denominator") if item.get("denominator") is not None else "—",
            "exactness": item.get("exactness", EXACTNESS),
        })
    display(pd.DataFrame(rows[:limit]))
    if len(rows) > limit:
        display(Markdown(f"*Displayed {limit} of {len(rows)} metrics; the complete machine payload is in `summary.json`.*"))

def interpretation(text):
    display(Markdown("**Interpretation.** " + text))

FRAME = None
if FRAME_URI:
    lowered = FRAME_URI.replace("\\", "/").lower()
    assert not any(marker in lowered for marker in ("/gold/", "gold_weather_z", "model_ready", "model-ready"))
    assert ("/eda/silver/" in lowered or not lowered.startswith("s3://")), FRAME_URI
    import pyarrow.parquet as pq
    FRAME = pq.read_table(FRAME_URI).to_pandas()

import html, io, re
from IPython.display import HTML
from leviathan.eda.reader_charts import (
    chart_scope_record as _reader_chart_scope_record,
    compute_chart_payloads as _compute_reader_chart_payloads,
    render_chart_payload as _render_reader_chart_payload,
)

PROFILE = SUMMARY.get("profile", {})
SECTIONS = PROFILE.get("sections", {})
OUTPUT_QUARANTINE = TABLE_NAME == "silver_model_predictions"
READER = {}
for _reader_payload in (PROVENANCE.get("reader"), SUMMARY.get("reader")):
    if isinstance(_reader_payload, dict):
        READER.update(_reader_payload)

FULL_DISPLAY_MAX_ROWS = 200
BOUNDED_PREVIEW_MAX_ROWS = 40
MAX_READER_CHARTS = 4
MIN_TREND_POINTS = 8
MIN_SCATTER_POINTS = 20
MIN_INSIGHTS = 3
MAX_INSIGHTS = 10

pd.set_option("display.max_rows", max(PREVIEW_ROWS, BOUNDED_PREVIEW_MAX_ROWS))
pd.set_option("display.max_columns", 40)
pd.set_option("display.max_colwidth", 120)
plt.rcParams.update({
    "axes.spines.top": False,
    "axes.spines.right": False,
    "grid.alpha": 0.18,
})

def first_value(*values, default=None):
    for value in values:
        if value is not None:
            return value
    return default

def as_list(value):
    if value is None:
        return []
    if isinstance(value, list):
        return value
    if isinstance(value, tuple):
        return list(value)
    return [value]

def text_items(value):
    items = []
    for item in as_list(value):
        if isinstance(item, dict):
            item = first_value(
                item.get("insight"), item.get("statement"), item.get("text"), item.get("message"),
                item.get("finding"), item.get("title"),
            )
        if item is not None and str(item).strip():
            items.append(str(item).strip())
    return items

def humanize(name):
    return re.sub(r"\s+", " ", str(name).replace("_", " ")).strip().capitalize()

def display_value(value):
    if value is None or (isinstance(value, float) and np.isnan(value)):
        return "not assessed"
    if isinstance(value, (np.integer, int)):
        return f"{int(value):,}"
    if isinstance(value, (np.floating, float)):
        return f"{float(value):,.3g}"
    return bounded(value, 180)

def rate_display(value):
    """Format rates without rounding a real, small issue down to zero."""
    if value is None or pd.isna(value):
        return "not assessed"
    rate = float(value)
    if rate == 0:
        return "0%"
    if abs(rate) < 0.0001:
        return f"{rate:.6%}"
    if abs(rate) < 0.01:
        return f"{rate:.4%}"
    return f"{rate:.2%}"

SOURCE_SHAPE = READER.get("source_shape", {})
if not isinstance(SOURCE_SHAPE, dict):
    SOURCE_SHAPE = {}
READER_SOURCE_SHAPE = SOURCE_SHAPE.get("source_shape")
READER_ANALYSIS_SHAPE = SOURCE_SHAPE.get("analysis_shape")
PROFILE_ROWS = metric_value("schema_contract", "row_count", None)
SOURCE_ROWS = int(first_value(
    READER.get("source_rows"), SOURCE_SHAPE.get("source_rows"), SOURCE_SHAPE.get("rows"),
    READER_SOURCE_SHAPE[0] if isinstance(READER_SOURCE_SHAPE, list) and READER_SOURCE_SHAPE else None,
    PROVENANCE.get("source_total_rows"), PROFILE_ROWS,
    len(FRAME) if FRAME is not None else None, default=0,
) or 0)
ANALYSIS_ROWS = int(first_value(
    READER.get("analysis_rows"), SOURCE_SHAPE.get("analysis_rows"), SOURCE_SHAPE.get("analyzed_rows"),
    READER_ANALYSIS_SHAPE[0] if isinstance(READER_ANALYSIS_SHAPE, list) and READER_ANALYSIS_SHAPE else None,
    PROFILE_ROWS, len(FRAME) if FRAME is not None else None, default=0,
) or 0)
ANALYSIS_COLUMNS = int(first_value(
    SOURCE_SHAPE.get("column_count"), SOURCE_SHAPE.get("columns"),
    READER_ANALYSIS_SHAPE[1] if isinstance(READER_ANALYSIS_SHAPE, list) and len(READER_ANALYSIS_SHAPE) > 1 else None,
    metric_value("schema_contract", "column_count", None),
    len(FRAME.columns) if FRAME is not None else None, default=0,
) or 0)

def declared_column_names():
    names = []
    declared = as_list(CONTRACT.get("physical_columns")) + as_list(CONTRACT.get("partition_keys"))
    for item in declared:
        if isinstance(item, dict) and item.get("name"):
            names.append(str(item["name"]))
    names.extend(str(name) for name in (metric_value("schema_contract", "observed_columns", []) or []))
    if FRAME is not None:
        names.extend(str(name) for name in FRAME.columns)
    return list(dict.fromkeys(names))

ALL_COLUMNS = declared_column_names()
VALUE_COLUMNS = [str(value) for value in as_list(CONTRACT.get("value_columns"))]

def row_meaning():
    configured = first_value(
        READER.get("row_meaning"), READER.get("grain"), OVERLAY.get("row_meaning"),
    )
    if isinstance(configured, dict):
        configured = first_value(configured.get("description"), configured.get("text"))
    if configured:
        return str(configured)
    natural_key = [str(value) for value in as_list(CONTRACT.get("natural_key"))]
    if natural_key:
        keys = ", ".join(f"`{name}`" for name in natural_key)
        return f"One row is identified by {keys}."
    return "The registry does not declare a complete natural key; inferred grain is evidence only."

def _stable_hash_order(frame):
    if frame.empty:
        return frame
    text = frame.astype("string").fillna("<NA>")
    order = pd.util.hash_pandas_object(text, index=False).sort_values(kind="mergesort").index
    return frame.loc[order]

def _parse_reader_time(series, name):
    if pd.api.types.is_datetime64_any_dtype(series):
        return pd.to_datetime(series, errors="coerce")
    if pd.api.types.is_numeric_dtype(series) and "year" in str(name).lower():
        years = pd.to_numeric(series, errors="coerce")
        text = years.round().astype("Int64").astype("string")
        return pd.to_datetime(text, format="%Y", errors="coerce")
    text = series.astype("string").str.strip()
    parsed = pd.Series(pd.NaT, index=series.index, dtype="datetime64[ns]")
    formats = (
        (text.str.match(r"^\d{4}-\d{2}-\d{2}(?:[ T].*)?$", na=False), "%Y-%m-%d", 10),
        (text.str.match(r"^\d{4}/\d{2}/\d{2}(?:[ T].*)?$", na=False), "%Y/%m/%d", 10),
        (text.str.match(r"^\d{8}$", na=False), "%Y%m%d", None),
        (text.str.match(r"^\d{4}-\d{2}$", na=False), "%Y-%m", None),
    )
    for mask, date_format, width in formats:
        if mask.any():
            values = text.loc[mask].str.slice(0, width) if width else text.loc[mask]
            parsed.loc[mask] = pd.to_datetime(values, format=date_format, errors="coerce")
    remaining = parsed.isna()
    years = text.loc[remaining].str.extract(r"((?:19|20)\d{2})", expand=False)
    parsed.loc[remaining] = pd.to_datetime(years, format="%Y", errors="coerce")
    return parsed

def _time_sort(frame):
    configured = as_list(READER.get("time_columns"))
    configured += as_list(READER.get("primary_time_column"))
    chart_plan = READER.get("chart_plan")
    if isinstance(chart_plan, dict):
        for chart in as_list(chart_plan.get("charts")):
            if isinstance(chart, dict):
                configured += as_list(chart.get("columns"))
    configured += as_list(CONTRACT.get("natural_key"))
    candidates = [str(value) for value in configured if str(value) in frame.columns]
    candidates += [
        str(column) for column in frame.columns
        if any(token in str(column).lower() for token in (
            "date", "year", "month", "week", "season", "time", "as_of",
        ))
    ]
    for column in dict.fromkeys(candidates):
        series = frame[column]
        parsed = _parse_reader_time(series, column)
        if parsed.notna().mean() >= 0.8:
            return column, parsed
    return None, None

def deterministic_preview(frame):
    if frame is None:
        return None, "No frozen analysis frame is attached to this notebook."
    if len(frame) <= FULL_DISPLAY_MAX_ROWS:
        return frame.copy(), f"Full DataFrame: all {len(frame):,} analyzed rows are shown."
    pieces = []
    time_column, order = _time_sort(frame)
    if time_column is not None:
        ordered = frame.assign(__reader_order=order).sort_values(
            "__reader_order", kind="mergesort", na_position="last",
        ).drop(columns="__reader_order")
        pieces.extend([ordered.head(10), ordered.tail(10)])
    entity_candidates = [
        str(value) for value in as_list(READER.get("preview_strata"))
        if str(value) in frame.columns and str(value) != time_column
        and frame[str(value)].nunique(dropna=True) <= 50
    ] + [
        str(value) for value in as_list(CONTRACT.get("natural_key"))
        if str(value) in frame.columns and str(value) != time_column
        and frame[str(value)].nunique(dropna=True) <= 50
    ]
    if entity_candidates:
        entity = entity_candidates[0]
        grouped = frame.sort_values(entity, kind="mergesort").groupby(entity, dropna=False)
        pieces.append(grouped.head(1).head(20))
    pieces.append(_stable_hash_order(frame).head(BOUNDED_PREVIEW_MAX_ROWS))
    preview = pd.concat(pieces, ignore_index=False)
    preview = preview.loc[~preview.index.duplicated()].drop_duplicates()
    preview = preview.head(BOUNDED_PREVIEW_MAX_ROWS)
    policy = (
        f"Deterministic preview: {len(preview):,} of {len(frame):,} analyzed rows; "
        "earliest/latest, entity examples, then stable-hash fill."
    )
    return preview, policy

def safe_preview(frame):
    configured = READER.get("preview")
    if isinstance(configured, dict) and isinstance(configured.get("records"), list):
        preview = pd.DataFrame(configured["records"])
        policy_value = READER.get("display_policy")
        if isinstance(policy_value, dict):
            policy_value = first_value(
                policy_value.get("description"), policy_value.get("mode"),
                default=bounded(policy_value, 500),
            )
        displayed_rows = configured.get("displayed_rows") or len(preview)
        policy = str(policy_value or (
            f"Reader evidence preview: {displayed_rows:,} "
            f"of {ANALYSIS_ROWS:,} analyzed rows; {configured.get('mode', 'bounded')}."
        ))
    else:
        preview, policy = deterministic_preview(frame)
    if preview is not None and OUTPUT_QUARANTINE:
        preview = preview.copy()
        for column in VALUE_COLUMNS:
            if column in preview:
                preview[column] = "[quarantined output value]"
        policy += " Output-plane value columns are masked."
    return preview, policy

def _configured_column_dictionary():
    configured = first_value(
        READER.get("column_dictionary"), READER.get("column_descriptions"),
        READER.get("columns"), default={},
    )
    if isinstance(configured, list):
        return {
            str(item.get("column") or item.get("name")): item
            for item in configured
            if isinstance(item, dict) and (item.get("column") or item.get("name"))
        }
    return configured if isinstance(configured, dict) else {}

def _column_description(column, configured):
    item = configured.get(column, {})
    configured_why = READER.get("why_it_matters", {})
    if not isinstance(configured_why, dict):
        configured_why = {}
    if isinstance(item, str):
        return item, str(
            configured_why.get(column) or "Supports interpretation and feature review."
        )
    if isinstance(item, dict):
        meaning = first_value(item.get("meaning"), item.get("description"), item.get("text"))
        why = first_value(
            item.get("why_it_matters"), item.get("feature_relevance"), item.get("role"),
        )
        if meaning:
            return str(meaning), str(
                why or configured_why.get(column)
                or "Supports interpretation and feature review."
            )
    lower = column.lower()
    if any(token in lower for token in ("date", "year", "month", "week", "season", "as_of")):
        return (
            f"{humanize(column)} used to place observations in time.",
            "Controls ordering, lagging, and cutoff safety.",
        )
    if column in VALUE_COLUMNS:
        return (
            f"Source measure: {humanize(column).lower()}.",
            "Potential feature input; units and timing must be governed.",
        )
    if lower == "source" or lower.endswith("_source"):
        return (
            "Source-system or release provenance label.",
            "Supports lineage and source consistency checks.",
        )
    if lower.endswith("_id") or column in as_list(CONTRACT.get("natural_key")):
        return (
            f"Identifier or grain field: {humanize(column).lower()}.",
            "Controls uniqueness and safe joins.",
        )
    return (
        f"Source field: {humanize(column).lower()}.",
        "Retained for contract, grouping, or quality analysis.",
    )

def column_dictionary():
    configured = _configured_column_dictionary()
    declared = as_list(CONTRACT.get("physical_columns")) + as_list(CONTRACT.get("partition_keys"))
    contract_columns = {
        str(item.get("name")): item
        for item in declared if isinstance(item, dict) and item.get("name")
    }
    missingness = metric_value("missingness_validity", "column_missingness", {}) or {}
    vocabularies = metric_value("entity_vocabulary_coverage", "vocabularies", {}) or {}
    distributions = metric_value("distributions", "numeric_distributions", {}) or {}
    rows = []
    for column in ALL_COLUMNS:
        declared_column = contract_columns.get(column, {})
        configured_item = configured.get(column, {})
        if not isinstance(configured_item, dict):
            configured_item = {}
        meaning, why = _column_description(column, configured)
        if configured_item and any(
            key in configured_item
            for key in ("observed_dtype", "declared_type", "null_rate", "distinct_count")
        ):
            dtype = first_value(
                configured_item.get("observed_dtype"), configured_item.get("declared_type"),
                default="not observed",
            )
            null_rate = configured_item.get("null_rate")
            distinct = configured_item.get("distinct_count")
            example = configured_item.get("example")
        elif FRAME is not None and column in FRAME:
            series = FRAME[column]
            dtype = str(series.dtype)
            null_rate = float(series.isna().mean()) if len(series) else None
            distinct = int(series.nunique(dropna=True))
            examples = series.dropna().head(1).tolist()
            example = examples[0] if examples else None
        else:
            dtype = first_value(
                declared_column.get("target_arrow_type"), declared_column.get("arrow_type"),
                declared_column.get("glue_type"), default="not observed",
            )
            null_rate = (missingness.get(column) or {}).get("null_rate")
            distinct = first_value(
                (vocabularies.get(column) or {}).get("distinct_count"),
                (distributions.get(column) or {}).get("distinct_count"),
            )
            example = None
        if OUTPUT_QUARANTINE and column in VALUE_COLUMNS:
            example = "[quarantined]"
        reader_units = READER.get("units", {})
        if not isinstance(reader_units, dict):
            reader_units = {}
        item = configured_item
        unit = first_value(
            item.get("unit") if isinstance(item, dict) else None,
            reader_units.get(column),
            OVERLAY.get("units", {}).get(column), default="not declared",
        )
        rows.append({
            "column": column,
            "dtype": dtype,
            "unit": unit,
            "example": bounded(example, 80) if example is not None else "not available",
            "null_rate": rate_display(null_rate) if ANALYSIS_ROWS else "not assessed",
            "distinct": distinct if ANALYSIS_ROWS else None,
            "plain_English_meaning": meaning,
            "why_it_matters": why,
        })
    return pd.DataFrame(rows)

def dashboard_kpis():
    configured = READER.get("dashboard_kpis")
    rows = []
    if isinstance(configured, dict):
        configured = [{"label": key, "value": value} for key, value in configured.items()]
    for item in as_list(configured):
        if isinstance(item, dict) and "value" in item:
            rows.append({
                "label": first_value(item.get("label"), item.get("name"), default="Metric"),
                "value": item.get("value"),
                "detail": first_value(item.get("detail"), item.get("unit"), default=""),
            })
    cap = SUMMARY.get("decision_capsule", {})
    defaults = [
        {"label": "Source shape", "value": f"{SOURCE_ROWS:,} x {ANALYSIS_COLUMNS:,}", "detail": EXACTNESS},
        {"label": "Analyzed rows", "value": ANALYSIS_ROWS, "detail": EXACTNESS},
        {"label": "Disposition", "value": (cap.get("disposition") or {}).get("value", PROFILE.get("disposition", "not assessed")), "detail": "feature readiness"},
        {"label": "PIT status", "value": "eligible" if metric_value("pit_leakage", "feature_eligible_source", False) else "not ready", "detail": "cutoff safety"},
        {"label": "Candidates", "value": len(SUMMARY.get("feature_candidates", [])), "detail": "review only"},
        {"label": "Blockers", "value": len(PROFILE.get("blockers", [])), "detail": "quality / contract"},
    ]
    existing = {str(row["label"]).lower() for row in rows}
    rows.extend(row for row in defaults if str(row["label"]).lower() not in existing)
    return rows[:8]

def show_kpi_cards(rows):
    cards = []
    for row in rows:
        cards.append(
            '<div style="border:1px solid #d9e0e8;border-radius:8px;padding:12px;min-width:145px;flex:1">'
            f'<div style="font-size:12px;color:#52606d">{html.escape(str(row["label"]))}</div>'
            f'<div style="font-size:22px;font-weight:650;color:#172b4d;margin:3px 0">{html.escape(display_value(row["value"]))}</div>'
            f'<div style="font-size:11px;color:#6b778c">{html.escape(str(row.get("detail") or ""))}</div>'
            '</div>'
        )
    markup = '<div style="display:flex;gap:8px;flex-wrap:wrap">' + ''.join(cards) + '</div>'
    display(HTML(markup))

def ordinary_describe():
    configured = first_value(
        READER.get("ordinary_statistics"), READER.get("descriptive_statistics")
    )
    if isinstance(configured, dict) and any(
        key in configured for key in ("numeric", "categorical", "temporal")
    ):
        frames = []
        for section_name in ("numeric", "categorical", "temporal"):
            section = configured.get(section_name) or {}
            records = section.get("records") if isinstance(section, dict) else None
            if isinstance(records, list) and records:
                frame = pd.DataFrame(records)
                frame.insert(0, "statistics_section", section_name)
                frames.append(frame)
        if frames:
            return pd.concat(frames, ignore_index=True, sort=False)
        return pd.DataFrame([{
            "status": configured.get("status", "not assessed"),
            "reason": "Reader statistics contain no displayable records.",
        }])
    if isinstance(configured, list) and configured:
        return pd.DataFrame(configured)
    if isinstance(configured, dict) and configured:
        return pd.DataFrame(configured).transpose()
    if FRAME is None or FRAME.empty:
        return pd.DataFrame([{"status": "not assessed", "reason": "No analysis frame attached."}])
    allowed = [
        column for column in FRAME.columns
        if not (OUTPUT_QUARANTINE and column in VALUE_COLUMNS)
    ]
    if not allowed:
        return pd.DataFrame([{"status": "not assessed", "reason": "Output values are quarantined."}])
    return FRAME[allowed].describe(include="all").transpose()

def quality_scorecard():
    configured = READER.get("quality_scorecard")
    if isinstance(configured, dict) and isinstance(configured.get("checks"), list):
        configured = configured["checks"]
    if isinstance(configured, dict):
        configured = [
            {"check": key, **(value if isinstance(value, dict) else {"status": value})}
            for key, value in configured.items()
        ]
    if isinstance(configured, list) and configured:
        rows = []
        for item in configured:
            if isinstance(item, dict):
                rows.append({
                    "check": first_value(item.get("check"), item.get("name"), default="Quality check"),
                    "status": str(item.get("status", "not assessed")).upper(),
                    "evidence": bounded(first_value(
                        item.get("summary"), item.get("evidence"), item.get("detail"),
                        default="",
                    ), 500),
                })
        if OUTPUT_QUARANTINE:
            for row in rows:
                if "validity" in str(row.get("check") or "").lower():
                    row["status"] = "NOT ASSESSED"
                    row["evidence"] = (
                        "Output-plane numeric value validity is quarantined; schema, null, "
                        "freshness, partition, and duplicate QA remain in scope."
                    )
        if rows:
            return pd.DataFrame(rows)
    checks = (
        "Grain and duplicates", "Schema contract", "Required-field completeness",
        "Temporal coverage / cadence", "Validity / identities", "PIT availability",
    )
    if ANALYSIS_ROWS <= 0:
        return pd.DataFrame([
            {"check": name, "status": "NOT ASSESSED", "evidence": "No analyzed rows; a zero rate is not inferred."}
            for name in checks
        ])
    duplicate_rate = metric_value("grain_integrity", "duplicate_key_row_rate", None)
    missing_required = metric_value("schema_contract", "missing_required_columns", []) or []
    required_rate = metric_value("missingness_validity", "required_complete_row_rate", None)
    required_complete_rows = metric_value(
        "missingness_validity", "required_complete_row_count", None
    )
    row_count = metric_value("schema_contract", "row_count", None)
    temporal = metric_value("temporal_structure", "usable_temporal_columns", []) or []
    pit_ready = bool(metric_value("pit_leakage", "feature_eligible_source", False))
    validity = None if OUTPUT_QUARANTINE else metric_value(
        "missingness_validity", "validity_summary", None
    )
    duplicate_evidence = (
        "No duplicate natural-key rows." if duplicate_rate == 0
        else f"Duplicate-key row rate: {duplicate_rate:.2%}." if duplicate_rate is not None
        else "Natural-key duplicate rate unavailable."
    )
    if required_rate is None:
        required_evidence = "Completeness denominator unavailable."
    elif row_count is not None and required_complete_rows is not None:
        incomplete_rows = max(0, int(row_count) - int(required_complete_rows))
        required_evidence = (
            f"{incomplete_rows:,} incomplete rows of {int(row_count):,} "
            f"({rate_display(incomplete_rows / int(row_count))} incomplete)."
            if int(row_count)
            else "No positive completeness denominator; a zero rate is not inferred."
        )
    else:
        required_evidence = f"Complete-row rate: {rate_display(required_rate)}."
    return pd.DataFrame([
        {"check": checks[0], "status": "PASS" if duplicate_rate == 0 else ("CAVEAT" if duplicate_rate is not None else "NOT ASSESSED"), "evidence": duplicate_evidence},
        {"check": checks[1], "status": "BLOCKER" if missing_required else "PASS", "evidence": "Missing required columns: " + ", ".join(missing_required) if missing_required else "Required columns are present."},
        {"check": checks[2], "status": "PASS" if required_rate == 1 else ("CAVEAT" if required_rate is not None else "NOT ASSESSED"), "evidence": required_evidence},
        {"check": checks[3], "status": "PASS" if temporal else "NOT ASSESSED", "evidence": "Usable axes: " + ", ".join(map(str, temporal)) if temporal else "No validated temporal axis."},
        {"check": checks[4], "status": "CAVEAT" if validity else "NOT ASSESSED", "evidence": bounded(validity, 500) if validity else "No source-specific identity rule supplied."},
        {"check": checks[5], "status": "PASS" if pit_ready else "BLOCKER", "evidence": "Source passed the source-level PIT gate." if pit_ready else "Knowledge-time availability is not safe enough for promotion."},
    ])

def computed_insights():
    configured = first_value(READER.get("reader_insights"), READER.get("insights"))
    if isinstance(configured, list) and any(isinstance(item, dict) for item in configured):
        # Reader-facing insights should be actual findings, not boilerplate
        # explaining why a configured analytical rule could not run.
        configured = [
            item for item in configured
            if not isinstance(item, dict)
            or str(item.get("status") or "evaluated").lower() == "evaluated"
        ]
    insights = text_items(configured)
    if not insights:
        insights.append(
            f"The evidence contains {ANALYSIS_ROWS:,} analyzed rows and {ANALYSIS_COLUMNS:,} columns; source coverage reports {SOURCE_ROWS:,} rows ({EXACTNESS})."
        )
        if FRAME is not None and not FRAME.empty:
            missing = FRAME.isna().mean().sort_values(ascending=False)
            if len(missing):
                column = str(missing.index[0])
                insights.append(
                    f"`{column}` has the highest observed null rate at {rate_display(missing.iloc[0])}; missing must not be read as zero."
                )
            time_column, order = _time_sort(FRAME)
            if time_column is not None and order.notna().any():
                valid = order.dropna()
                insights.append(
                    f"`{time_column}` spans {valid.min()} to {valid.max()} across {valid.nunique():,} observed periods."
                )
        duplicate_rate = metric_value("grain_integrity", "duplicate_key_row_rate", None)
        if duplicate_rate is not None and ANALYSIS_ROWS:
            insights.append(
                f"Declared-key duplicate-row rate is {duplicate_rate:.2%} across the analyzed frame."
            )
        insights.append(
            "The source-level PIT gate passed; each transformation still needs a cutoff rule."
            if metric_value("pit_leakage", "feature_eligible_source", False)
            else "The source is not PIT-ready; final/current values must not be treated as known earlier."
        )
    fallbacks = [
        f"There are {len(SUMMARY.get('feature_candidates', []))} review-only candidate(s); none is enabled here.",
        f"The registry declares {len(VALUE_COLUMNS)} feature-bearing value column(s).",
        "The analysis is source-only; no target association was calculated.",
    ]
    for item in fallbacks:
        if len(insights) >= MIN_INSIGHTS:
            break
        insights.append(item)
    # Five concise observations are enough to understand the source for feature
    # engineering; detailed evidence remains in the tables and sidecars.
    return list(dict.fromkeys(insights))[:min(MAX_INSIGHTS, 5)]

def notable_observations():
    configured = first_value(READER.get("notable_observations"), READER.get("notable_rows"))
    if isinstance(configured, dict):
        records = first_value(configured.get("records"), configured.get("rows"))
        if isinstance(records, list):
            flattened = []
            for item in records:
                if not isinstance(item, dict):
                    continue
                row = item.get("row") if isinstance(item.get("row"), dict) else item
                flattened.append({
                    "why_notable": "; ".join(map(str, as_list(item.get("reasons")))),
                    **row,
                })
            configured = flattened
        else:
            configured = []
    if isinstance(configured, list) and configured:
        return pd.DataFrame(configured).head(12)
    if FRAME is None or FRAME.empty:
        return pd.DataFrame([{"status": "not assessed", "reason": "No analyzed rows."}])
    if OUTPUT_QUARANTINE:
        preview, _ = safe_preview(FRAME)
        return preview.head(4) if preview is not None else pd.DataFrame()
    time_column, order = _time_sort(FRAME)
    selected, labels = [], []
    if time_column is not None and order.notna().any():
        valid_indices = order.dropna().sort_values(kind="mergesort").index
        selected.extend([valid_indices[0], valid_indices[-1]])
        labels.extend(["earliest observed row", "latest observed row"])
    measures = [
        column for column in VALUE_COLUMNS
        if column in FRAME and pd.api.types.is_numeric_dtype(FRAME[column])
    ]
    for column in measures[:3]:
        values = pd.to_numeric(FRAME[column], errors="coerce")
        if values.notna().any():
            selected.extend([values.idxmin(), values.idxmax()])
            labels.extend([f"lowest {column}", f"highest {column}"])
    if not selected:
        selected = list(_stable_hash_order(FRAME).head(4).index)
        labels = ["deterministic example"] * len(selected)
    rows = FRAME.loc[selected].copy().reset_index(drop=True)
    rows.insert(0, "why_notable", labels)
    return rows.drop_duplicates().head(10)

def _measure_columns(frame):
    entity_axes = {
        "commodity_code", "country_code", "entity_id", "source_unit_id",
        "state_code", "region_code", "series_id",
    }
    time_tokens = (
        "date", "year", "month", "week", "season", "period", "as_of",
        "time", "survey_number", "fortnight_seq",
    )
    def eligible(column):
        name = str(column)
        lower = name.lower()
        return (
            name in frame.columns
            and pd.api.types.is_numeric_dtype(frame[name])
            and lower not in entity_axes
            and not lower.endswith(("_id", "_code"))
            and not any(token in lower for token in time_tokens)
        )
    preferred = [
        column
        for column in text_items(READER.get("primary_measures")) + VALUE_COLUMNS
        if eligible(column)
    ]
    numeric = [
        str(column)
        for column in frame.select_dtypes(include=[np.number]).columns
        if eligible(column)
    ]
    return list(dict.fromkeys(preferred + numeric))[:MAX_NUMERIC_CHARTS]

def _chart_scope(plotted_rows, chart_exactness, scope):
    return (
        f"source n={SOURCE_ROWS:,}; analyzed n={ANALYSIS_ROWS:,}; plotted n={plotted_rows:,}; "
        f"exactness={chart_exactness}"
    )

def _scope_prose(scope):
    if isinstance(scope, dict):
        skip = {"analysis_rows", "exactness", "plotted_rows", "source_rows"}
        parts = []
        for key, value in scope.items():
            if key in skip or value in (None, "", [], {}):
                continue
            if isinstance(value, (dict, list, tuple)):
                continue
            parts.append(f"{humanize(key)}={value}")
        return "; ".join(parts) or "complete canonical source aggregate"
    return bounded(scope or "complete canonical source aggregate", 500)

def _chart_unit(measure, configured):
    unit = str(configured or "").strip()
    if unit and "verify" not in unit.lower() and "source-native" not in unit.lower():
        return unit
    name = str(measure).lower()
    suffixes = {
        "_mm": "mm", "_pct": "%", "_percent": "%", "_mt": "metric tonnes",
        "_kg": "kg", "_ha": "hectares", "_c": "°C", "_kt": "thousand tonnes",
    }
    return next((label for suffix, label in suffixes.items() if name.endswith(suffix)), "unit not declared")

def _aggregate_footer(fig, plotted_rows, scope):
    footer = _chart_scope(plotted_rows, "exact full-source aggregate", scope)
    fig.text(0.015, 0.012, footer, fontsize=7.5, color="#52606d", va="bottom")
    fig.tight_layout(rect=(0.01, 0.18, 0.99, 0.97))

def reader_chart_plans():
    plans = READER.get("chart_plan")
    if isinstance(plans, dict):
        plans = plans.get("charts", [])
    return [item for item in as_list(plans) if isinstance(item, dict)]

def _source_aggregate_charts():
    source_aggregates = READER.get("source_aggregates", {})
    if not isinstance(source_aggregates, dict):
        return [], []
    chart_tables = source_aggregates.get("chart_tables")
    if not chart_tables:
        return [], []
    plans = reader_chart_plans()
    plans_by_table = {
        str(first_value(
            item.get("table"), item.get("table_id"), item.get("chart_id"), item.get("id")
        )): item
        for item in plans
        if first_value(
            item.get("table"), item.get("table_id"), item.get("chart_id"), item.get("id")
        )
    }
    if isinstance(chart_tables, dict):
        entries = []
        for table_id, value in chart_tables.items():
            if isinstance(value, dict):
                entries.append({"table_id": table_id, **value})
            else:
                entries.append({"table_id": table_id, "rows": value})
    else:
        entries = [item for item in as_list(chart_tables) if isinstance(item, dict)]
    chart_priority = [str(value) for value in as_list(source_aggregates.get("chart_priority"))]
    if chart_priority:
        priority_rank = {table_id: index for index, table_id in enumerate(chart_priority)}
        entries.sort(
            key=lambda entry: (
                priority_rank.get(
                    str(first_value(entry.get("table_id"), entry.get("id"), default="")),
                    len(priority_rank),
                ),
                str(first_value(entry.get("table_id"), entry.get("id"), default="")),
            )
        )
    records, reasons = [], []
    for entry in entries[:MAX_READER_CHARTS]:
        table_id = str(first_value(entry.get("table_id"), entry.get("id"), default="chart"))
        plan = {**plans_by_table.get(table_id, {}), **entry}
        raw_rows = first_value(plan.get("rows"), plan.get("records"), plan.get("data"))
        if isinstance(raw_rows, dict):
            chart_frame = pd.DataFrame(raw_rows)
        elif isinstance(raw_rows, list):
            chart_frame = pd.DataFrame(raw_rows)
        else:
            reasons.append(f"Exact aggregate `{table_id}` has no tabular rows.")
            continue
        if chart_frame.empty:
            reasons.append(f"Exact aggregate `{table_id}` is empty.")
            continue
        kind = str(first_value(plan.get("kind"), plan.get("type"), default="line")).lower()
        columns = [str(column) for column in as_list(plan.get("columns"))]
        x = first_value(plan.get("x"), plan.get("x_column"), columns[0] if columns else None)
        y = first_value(
            plan.get("y"), plan.get("value_column"),
            columns[1] if len(columns) > 1 else None,
        )
        series = first_value(plan.get("series"), plan.get("group"), plan.get("color"))
        numeric = list(chart_frame.select_dtypes(include=[np.number]).columns)
        if y is None and numeric:
            y = numeric[-1]
        if x is None:
            candidates = [column for column in chart_frame.columns if column != y]
            x = candidates[0] if candidates else None
        if x not in chart_frame or y not in chart_frame:
            reasons.append(f"Exact aggregate `{table_id}` lacks usable x/y fields.")
            continue
        title = str(plan.get("title") or humanize(table_id))
        if re.search(r"\bvalue\b", title, flags=re.IGNORECASE):
            title = re.sub(r"\bvalue\b", humanize(str(y)), title, flags=re.IGNORECASE)
        scope = _scope_prose(plan.get("scope"))
        unit = _chart_unit(y, plan.get("unit") or OVERLAY.get("units", {}).get(y))
        plotted_rows = len(chart_frame)
        fig, ax = plt.subplots(figsize=(9, 4.8))
        if kind in {"line", "area"}:
            if chart_frame[x].nunique(dropna=True) < MIN_TREND_POINTS:
                plt.close(fig)
                reasons.append(
                    f"No exact `{table_id}` trend: fewer than {MIN_TREND_POINTS} ordered points."
                )
                continue
            if series in chart_frame:
                for label, group in chart_frame.groupby(series, dropna=False):
                    ordered = group.sort_values(x, kind="mergesort")
                    ax.plot(ordered[x], ordered[y], marker="o", markersize=2.5, label=str(label))
                if chart_frame[series].nunique(dropna=False) > 1:
                    ax.legend(frameon=False)
            else:
                ordered = chart_frame.sort_values(x, kind="mergesort")
                ax.plot(ordered[x], ordered[y], marker="o", markersize=2.5, color="#356AA0")
            if "year" in str(x).lower() and any(token in str(y).lower() for token in ("count", "row", "coverage")):
                years = pd.to_numeric(chart_frame[x], errors="coerce")
                values = pd.to_numeric(chart_frame[y], errors="coerce")
                valid = pd.DataFrame({"year": years, "value": values}).dropna().sort_values("year")
                if not valid.empty and int(valid.iloc[-1]["year"]) == pd.Timestamp.utcnow().year:
                    latest = valid.iloc[-1]
                    ax.annotate(
                        "partial year",
                        (latest["year"], latest["value"]),
                        xytext=(6, 8), textcoords="offset points", fontsize=8, color="#D1495B",
                    )
                    title += " (latest year is partial)"
                    scope += "; latest calendar year is partial"
        elif kind in {"bar", "ranking", "histogram"}:
            shown = chart_frame.head(20)
            plotted_rows = len(shown)
            labels = shown[x].astype("string")
            values = pd.to_numeric(shown[y], errors="coerce")
            if labels.str.len().max() > 12 or len(shown) > 8:
                ax.barh(labels, values, color="#356AA0")
                ax.set_xlabel(unit); ax.set_ylabel(str(x))
            else:
                ax.bar(labels, values, color="#356AA0")
                ax.set_ylabel(unit); ax.set_xlabel(str(x))
                ax.tick_params(axis="x", rotation=35)
        elif kind == "scatter":
            points = chart_frame[[x, y]].apply(pd.to_numeric, errors="coerce").dropna()
            if len(points) < MIN_SCATTER_POINTS:
                plt.close(fig)
                reasons.append(
                    f"No exact `{table_id}` scatter: only {len(points)} comparable points; "
                    f"at least {MIN_SCATTER_POINTS} are required."
                )
                continue
            ax.scatter(points[x], points[y], alpha=0.6, color="#356AA0", edgecolors="none")
            plotted_rows = len(points)
        elif kind == "heatmap":
            value = first_value(plan.get("value"), plan.get("value_column"), y)
            row = first_value(plan.get("row"), plan.get("y_axis"))
            column = first_value(plan.get("column"), plan.get("x_axis"), x)
            if row not in chart_frame or column not in chart_frame or value not in chart_frame:
                plt.close(fig)
                reasons.append(f"Exact aggregate `{table_id}` lacks heatmap row/column/value fields.")
                continue
            matrix = chart_frame.pivot_table(index=row, columns=column, values=value, aggfunc="first")
            is_anomaly = "anomaly" in table_id.lower() or "z-score" in unit.lower()
            if is_anomaly:
                finite = np.asarray(matrix.values, dtype=float)
                finite = finite[np.isfinite(finite)]
                bound = float(np.max(np.abs(finite))) if len(finite) else 1.0
                bound = bound if bound > 0 else 1.0
                image = ax.imshow(
                    matrix.values,
                    aspect="auto",
                    cmap="RdBu_r",
                    vmin=-bound,
                    vmax=bound,
                )
            else:
                image = ax.imshow(matrix.values, aspect="auto", cmap="Blues")
            x_ticks = np.unique(np.linspace(0, max(0, len(matrix.columns) - 1), min(8, len(matrix.columns))).astype(int))
            y_ticks = np.unique(np.linspace(0, max(0, len(matrix.index) - 1), min(8, len(matrix.index))).astype(int))
            ax.set_xticks(x_ticks, [matrix.columns[index] for index in x_ticks], rotation=45, ha="right")
            ax.set_yticks(y_ticks, [matrix.index[index] for index in y_ticks])
            fig.colorbar(image, ax=ax, label=unit)
        else:
            plt.close(fig)
            reasons.append(f"Exact aggregate `{table_id}` requests unsupported chart kind `{kind}`.")
            continue
        ax.set_title(title)
        if kind not in {"bar", "ranking", "histogram", "heatmap"}:
            ax.set_xlabel(str(x)); ax.set_ylabel(unit if kind != "scatter" else str(y))
        _aggregate_footer(fig, plotted_rows, scope)
        plt.show()
        records.append({
            "chart": table_id,
            "title": title,
            "plotted_rows": plotted_rows,
            "source_rows": SOURCE_ROWS,
            "analysis_rows": ANALYSIS_ROWS,
            "exactness": "exact full-source aggregate",
            "scope": scope,
            "unit": unit,
            "aggregation": "exact aggregate rows rendered without notebook recomputation",
        })
    return records, reasons

def _legacy_planned_frame_charts(frame):
    records, reasons = [], []
    for plan in reader_chart_plans():
        if len(records) >= MAX_READER_CHARTS:
            break
        status = str(plan.get("status") or "ready").lower()
        if status not in {"ready", "supported", "pass", "available"}:
            reasons.append(str(plan.get("reason") or f"Planned chart `{plan.get('chart_id')}` is {status}."))
            continue
        columns = [str(column) for column in as_list(plan.get("columns"))]
        columns = [column for column in columns if column in frame.columns]
        if not columns:
            reasons.append(f"Planned chart `{plan.get('chart_id')}` has no available columns.")
            continue
        raw_kind = str(first_value(
            plan.get("chart_type"), plan.get("type"), plan.get("kind"), default=""
        )).lower()
        if any(token in raw_kind for token in ("line", "trend", "curve", "revision", "time_series")):
            kind = "line"
        elif any(token in raw_kind for token in ("bar", "rank")):
            kind = "bar"
        elif "heat" in raw_kind:
            kind = "heatmap"
        elif any(token in raw_kind for token in ("hist", "distribution")):
            kind = "histogram"
        elif any(token in raw_kind for token in ("scatter", "relationship")):
            kind = "scatter"
        elif "box" in raw_kind:
            kind = "boxplot"
        else:
            reasons.append(
                f"Planned chart `{plan.get('chart_id')}` uses unsupported type `{raw_kind}`."
            )
            continue
        numeric = [column for column in columns if pd.api.types.is_numeric_dtype(frame[column])]
        non_numeric = [column for column in columns if column not in numeric]
        minimum_rows = int(plan.get("minimum_rows") or 1)
        title = str(plan.get("title") or humanize(plan.get("chart_id") or raw_kind))
        scope = bounded(plan.get("scope") or "frozen analysis frame", 300)
        fig, ax = plt.subplots(figsize=(9, 4.2), constrained_layout=True)
        plotted_rows = 0
        if kind == "line":
            time_column, time_order = _time_sort(frame[columns])
            y_columns = [column for column in numeric if column != time_column][:3]
            required = max(MIN_TREND_POINTS, minimum_rows)
            if time_column is None or not y_columns or time_order.nunique(dropna=True) < required:
                plt.close(fig)
                reasons.append(
                    f"No `{title}` line: fewer than {required} ordered periods or no numeric measure."
                )
                continue
            work = frame[[time_column] + y_columns].copy()
            work["__order"] = time_order
            for column in y_columns:
                work[column] = pd.to_numeric(work[column], errors="coerce")
            work = work.dropna(subset=["__order"])
            plotted = work.groupby("__order", as_index=False)[y_columns].median(numeric_only=True)
            for column in y_columns:
                ax.plot(plotted["__order"], plotted[column], marker="o", markersize=2.5, label=column)
            if len(y_columns) > 1:
                ax.legend(frameon=False, ncol=min(3, len(y_columns)))
            ax.set_xlabel(time_column)
            ax.set_ylabel("Source-native units; see column dictionary")
            plotted_rows = len(plotted)
        elif kind == "bar":
            category = non_numeric[0] if non_numeric else columns[0]
            measure = next((column for column in numeric if column != category), None)
            if measure:
                plotted = frame.groupby(category, dropna=False)[measure].median().sort_values().tail(15)
                label = OVERLAY.get("units", {}).get(measure, measure)
            else:
                plotted = frame[category].astype("string").fillna("<missing>").value_counts().head(15).sort_values()
                label = "Rows"
            if len(plotted) < max(2, minimum_rows):
                plt.close(fig)
                reasons.append(f"No `{title}` bar: too few comparable categories.")
                continue
            plotted.plot.barh(ax=ax, color="#356AA0", legend=False)
            ax.set_xlabel(label); ax.set_ylabel(category)
            plotted_rows = len(plotted)
        elif kind == "histogram":
            measure = numeric[0] if numeric else None
            values = (
                pd.to_numeric(frame[measure], errors="coerce").dropna()
                if measure else pd.Series(dtype=float)
            )
            required = max(MIN_TREND_POINTS, minimum_rows)
            if len(values) < required or values.nunique() < 4:
                plt.close(fig)
                reasons.append(f"No `{title}` distribution: insufficient populated variation.")
                continue
            bins = min(40, max(5, int(np.sqrt(len(values)))))
            ax.hist(values, bins=bins, color="#356AA0", edgecolor="white")
            ax.set_xlabel(OVERLAY.get("units", {}).get(measure, measure)); ax.set_ylabel("Rows")
            plotted_rows = len(values)
        elif kind == "scatter":
            if len(numeric) < 2:
                plt.close(fig)
                reasons.append(f"No `{title}` scatter: two numeric columns are required.")
                continue
            x, y = numeric[:2]
            points = frame[[x, y]].apply(pd.to_numeric, errors="coerce").dropna()
            required = max(MIN_SCATTER_POINTS, minimum_rows)
            if len(points) < required:
                plt.close(fig)
                reasons.append(
                    f"No `{title}` scatter: only {len(points)} comparable rows; {required} required."
                )
                continue
            ax.scatter(points[x], points[y], alpha=0.55, color="#356AA0", edgecolors="none")
            ax.set_xlabel(x); ax.set_ylabel(y)
            plotted_rows = len(points)
        elif kind == "heatmap":
            if len(columns) < 3 or not numeric:
                plt.close(fig)
                reasons.append(f"No `{title}` heatmap: row, column, and numeric value are required.")
                continue
            value = numeric[-1]
            axes = [column for column in columns if column != value][:2]
            matrix = frame.pivot_table(
                index=axes[0], columns=axes[1], values=value, aggfunc="mean"
            )
            if matrix.empty:
                plt.close(fig)
                reasons.append(f"No `{title}` heatmap: pivot is empty.")
                continue
            image = ax.imshow(matrix.values, aspect="auto", cmap="Blues")
            ax.set_xticks(range(len(matrix.columns)), matrix.columns, rotation=45, ha="right")
            ax.set_yticks(range(len(matrix.index)), matrix.index)
            fig.colorbar(image, ax=ax, label=OVERLAY.get("units", {}).get(value, value))
            plotted_rows = int(matrix.notna().sum().sum())
        else:
            measure = numeric[0] if numeric else None
            values = pd.to_numeric(frame[measure], errors="coerce").dropna() if measure else pd.Series(dtype=float)
            if len(values) < max(4, minimum_rows):
                plt.close(fig)
                reasons.append(f"No `{title}` boxplot: insufficient populated rows.")
                continue
            ax.boxplot(values, orientation="vertical", showfliers=True)
            ax.set_ylabel(OVERLAY.get("units", {}).get(measure, measure))
            plotted_rows = len(values)
        ax.set_title(title)
        fig.text(
            0.01, 0.01,
            _chart_scope(plotted_rows, EXACTNESS, scope),
            fontsize=8, color="#52606d",
        )
        plt.show()
        records.append({
            "chart": str(plan.get("chart_id") or plan.get("id") or title),
            "plotted_rows": plotted_rows,
            "exactness": EXACTNESS,
        })
    return records, reasons

def _planned_frame_charts(frame):
    payloads = _compute_reader_chart_payloads(
        frame,
        reader_chart_plans(),
        {
            **PROVENANCE,
            "source_rows": SOURCE_ROWS,
            "analysis_rows": ANALYSIS_ROWS,
            "exactness": EXACTNESS,
        },
        units=OVERLAY.get("units", {}),
        relationship_evidence=PROVENANCE.get("relationship_checks", []),
        quarantine=OUTPUT_QUARANTINE,
        max_charts=MAX_READER_CHARTS,
    )
    records, reasons, manifest = [], [], []
    for payload in payloads:
        if payload.get("status") != "ready":
            reason = payload.get("omission_reason") or "no honest payload is available."
            reasons.append(f"Planned chart `{payload.get('chart_id')}` omitted: {reason}")
            manifest.append({
                "chart_id": payload.get("chart_id"),
                "status": "omitted",
                "reason": reason,
                "source_rows": int(payload.get("source_rows") or SOURCE_ROWS or 0),
                "analysis_rows": int(payload.get("analysis_rows") or ANALYSIS_ROWS or 0),
                "plotted_rows": 0,
                "scope": payload.get("scope") or "frozen analysis frame",
                "exactness": payload.get("exactness") or EXACTNESS,
                "title": payload.get("title"),
                "unit": None,
                "aggregation": None,
            })
            continue
        figure = _render_reader_chart_payload(payload)
        if figure is None:
            reason = "renderer returned no figure."
            reasons.append(f"Planned chart `{payload.get('chart_id')}` omitted: {reason}")
            manifest.append({
                "chart_id": payload.get("chart_id"),
                "status": "omitted",
                "reason": reason,
                "source_rows": int(payload.get("source_rows") or SOURCE_ROWS or 0),
                "analysis_rows": int(payload.get("analysis_rows") or ANALYSIS_ROWS or 0),
                "plotted_rows": 0,
                "scope": payload.get("scope") or "frozen analysis frame",
                "exactness": payload.get("exactness") or EXACTNESS,
                "title": payload.get("title"),
                "unit": None,
                "aggregation": None,
            })
            continue
        display(figure)
        plt.close(figure)
        records.append(_reader_chart_scope_record(payload))
        manifest.append({
            "chart_id": payload.get("chart_id"),
            "status": "ready",
            "reason": None,
            "source_rows": int(payload.get("source_rows") or SOURCE_ROWS or 0),
            "analysis_rows": int(payload.get("analysis_rows") or ANALYSIS_ROWS or 0),
            "plotted_rows": int(payload.get("plotted_rows") or 0),
            "scope": payload.get("scope") or "frozen analysis frame",
            "exactness": payload.get("exactness") or EXACTNESS,
            "title": payload.get("title"),
            "unit": payload.get("unit"),
            "aggregation": payload.get("aggregation"),
        })
    return records, reasons, manifest

def _runtime_chart_manifest(records, reasons, *, prefix):
    manifest = [{
        "chart_id": record.get("chart") or f"{prefix}_{index + 1}",
        "status": "ready",
        "reason": None,
        "source_rows": int(record.get("source_rows") or SOURCE_ROWS or 0),
        "analysis_rows": int(record.get("analysis_rows") or ANALYSIS_ROWS or 0),
        "plotted_rows": int(record.get("plotted_rows") or 0),
        "scope": record.get("scope") or "reader runtime chart",
        "exactness": record.get("exactness") or EXACTNESS,
        "title": record.get("title") or record.get("chart"),
        "unit": record.get("unit"),
        "aggregation": record.get("aggregation"),
    } for index, record in enumerate(records)]
    for index, reason in enumerate(reasons):
        match = re.search(r"`([^`]+)`", str(reason))
        manifest.append({
            "chart_id": match.group(1) if match else f"{prefix}_omitted_{index + 1}",
            "status": "omitted",
            "reason": str(reason),
            "source_rows": int(SOURCE_ROWS or 0),
            "analysis_rows": int(ANALYSIS_ROWS or 0),
            "plotted_rows": 0,
            "scope": "runtime omission from approved reader evidence",
            "exactness": EXACTNESS,
            "title": None,
            "unit": None,
            "aggregation": None,
        })
    return manifest

def adaptive_charts():
    if OUTPUT_QUARANTINE:
        reason = "Model predictions are an output plane; value charts and relationships are quarantined."
        return [], [reason], [{
            "chart_id": "output_quarantine",
            "status": "omitted",
            "reason": reason,
            "source_rows": int(SOURCE_ROWS or 0),
            "analysis_rows": int(ANALYSIS_ROWS or 0),
            "plotted_rows": 0,
            "scope": "generated-output leakage quarantine",
            "exactness": EXACTNESS,
            "title": "Generated-output leakage quarantine",
            "unit": None,
            "aggregation": None,
        }]
    records, reasons = _source_aggregate_charts()
    if records:
        return records, reasons, _runtime_chart_manifest(records, reasons, prefix="source_aggregate")
    if FRAME is None or FRAME.empty:
        reasons.append("No frozen analysis frame is attached, so an honest chart cannot be drawn.")
        return records, reasons, _runtime_chart_manifest(records, reasons, prefix="frame")
    frame = FRAME
    planned = reader_chart_plans()
    records, planned_reasons, manifest = _planned_frame_charts(frame)
    reasons.extend(planned_reasons)
    if records:
        return records, reasons, manifest
    if planned:
        return records, reasons, manifest
    measures = _measure_columns(frame)
    time_column, time_order = _time_sort(frame)

    if (
        time_column is not None and measures
        and time_order.notna().sum() >= MIN_TREND_POINTS
        and time_order.nunique() >= MIN_TREND_POINTS
    ):
        selected = [column for column in measures if column != time_column][:3]
        plot_columns = list(dict.fromkeys([time_column, *selected]))
        plot_frame = frame[plot_columns].copy()
        plot_frame["__order"] = time_order
        plot_frame = plot_frame.dropna(subset=["__order"])
        for column in selected:
            plot_frame[column] = pd.to_numeric(plot_frame[column], errors="coerce")
        plotted = plot_frame.groupby("__order", as_index=False)[selected].median(numeric_only=True)
        plotted = plotted.dropna(how="all", subset=selected)
        if len(plotted) >= MIN_TREND_POINTS:
            chart_exactness = EXACTNESS
            if len(plotted) > CHART_MAX_POINTS:
                positions = np.linspace(0, len(plotted) - 1, CHART_MAX_POINTS, dtype=int)
                plotted = plotted.iloc[positions]
                chart_exactness = "sampled"
            fig, ax = plt.subplots(figsize=(10, 4.2), constrained_layout=True)
            for column in selected:
                ax.plot(
                    plotted["__order"], plotted[column], marker="o", markersize=2.5,
                    label=column,
                )
            ax.set_title(f"Selected measures over {time_column}")
            ax.set_xlabel(time_column)
            ax.set_ylabel("Source-native units; see column dictionary")
            if len(selected) > 1:
                ax.legend(frameon=False, ncol=min(3, len(selected)))
            fig.text(
                0.01, 0.01,
                _chart_scope(len(plotted), chart_exactness, "full time range"),
                fontsize=8, color="#52606d",
            )
            plt.show()
            records.append({
                "chart": "temporal measures", "plotted_rows": len(plotted),
                "exactness": chart_exactness,
            })
    elif time_column is not None:
        reasons.append(
            f"No line chart: `{time_column}` has fewer than {MIN_TREND_POINTS} comparable periods or no numeric measure."
        )

    if measures and len(records) < MAX_READER_CHARTS:
        column = measures[0]
        values = pd.to_numeric(frame[column], errors="coerce")
        values = values.replace([np.inf, -np.inf], np.nan).dropna()
        if len(values) >= MIN_TREND_POINTS and values.nunique() >= 4:
            plotted = values
            chart_exactness = EXACTNESS
            if len(plotted) > CHART_MAX_POINTS:
                positions = np.linspace(0, len(plotted) - 1, CHART_MAX_POINTS, dtype=int)
                plotted = plotted.iloc[positions]
                chart_exactness = "sampled"
            fig, ax = plt.subplots(figsize=(8.5, 3.8), constrained_layout=True)
            bins = min(40, max(5, int(np.sqrt(len(plotted)))))
            ax.hist(plotted, bins=bins, color="#356AA0", edgecolor="white")
            ax.set_title(f"Distribution of {column}")
            ax.set_xlabel(OVERLAY.get("units", {}).get(column, "source-native unit"))
            ax.set_ylabel("Rows")
            fig.text(
                0.01, 0.01,
                _chart_scope(len(plotted), chart_exactness, "all entities and periods"),
                fontsize=8, color="#52606d",
            )
            plt.show()
            records.append({
                "chart": f"distribution: {column}", "plotted_rows": len(plotted),
                "exactness": chart_exactness,
            })
        else:
            reasons.append(
                f"No distribution chart for `{column}`: fewer than {MIN_TREND_POINTS} populated observations or four distinct values."
            )

    categories = [
        str(column) for column in frame.columns
        if not pd.api.types.is_numeric_dtype(frame[column])
        and 2 <= frame[column].nunique(dropna=True) <= 20
        and str(column) != time_column
    ]
    if categories and len(records) < MAX_READER_CHARTS:
        column = categories[0]
        counts = frame[column].astype("string").fillna("<missing>")
        counts = counts.value_counts().head(12).sort_values()
        if len(counts) >= 2:
            fig, ax = plt.subplots(
                figsize=(8.5, max(3.2, len(counts) * 0.35)), constrained_layout=True,
            )
            counts.plot.barh(ax=ax, color="#356AA0", legend=False)
            ax.set_title(f"Most common {column} values")
            ax.set_xlabel("Rows"); ax.set_ylabel(column)
            fig.text(
                0.01, 0.01,
                _chart_scope(int(counts.sum()), EXACTNESS, "top 12 categories"),
                fontsize=8, color="#52606d",
            )
            plt.show()
            records.append({
                "chart": f"category coverage: {column}",
                "plotted_rows": int(counts.sum()), "exactness": EXACTNESS,
            })

    missing = frame.isna().mean().sort_values(ascending=False)
    missing = missing[missing.gt(0)].head(15).sort_values()
    if len(missing) and len(records) < MAX_READER_CHARTS:
        fig, ax = plt.subplots(
            figsize=(8.5, max(3.0, len(missing) * 0.32)), constrained_layout=True,
        )
        missing.plot.barh(ax=ax, color="#D99A2B", legend=False)
        ax.set_title("Columns with observed missing values")
        ax.set_xlabel("Share of analyzed rows"); ax.set_xlim(0, 1)
        fig.text(
            0.01, 0.01,
            _chart_scope(len(frame), EXACTNESS, "all rows; top 15 columns"),
            fontsize=8, color="#52606d",
        )
        plt.show()
        records.append({
            "chart": "missingness", "plotted_rows": len(frame), "exactness": EXACTNESS,
        })

    planned = reader_chart_plans()
    for plan in planned:
        if len(records) >= MAX_READER_CHARTS:
            break
        kind = str(first_value(plan.get("kind"), plan.get("type"), default="")).lower()
        if kind != "scatter":
            continue
        columns = [str(column) for column in as_list(plan.get("columns"))]
        x = first_value(plan.get("x"), columns[0] if columns else None)
        y = first_value(plan.get("y"), columns[1] if len(columns) > 1 else None)
        if x not in frame or y not in frame:
            reasons.append(f"No planned scatter: `{x}` or `{y}` is absent.")
            continue
        points = frame[[x, y]].apply(pd.to_numeric, errors="coerce").dropna()
        if len(points) < MIN_SCATTER_POINTS:
            reasons.append(
                f"No `{x}` versus `{y}` scatter: only {len(points)} comparable observations; "
                f"at least {MIN_SCATTER_POINTS} are required."
            )
            continue
        plotted = points
        chart_exactness = EXACTNESS
        if len(plotted) > CHART_MAX_POINTS:
            positions = np.linspace(0, len(plotted) - 1, CHART_MAX_POINTS, dtype=int)
            plotted = plotted.iloc[positions]
            chart_exactness = "sampled"
        fig, ax = plt.subplots(figsize=(6.5, 4.5), constrained_layout=True)
        ax.scatter(plotted[x], plotted[y], alpha=0.55, color="#356AA0", edgecolors="none")
        ax.set_title(str(plan.get("title") or f"{y} versus {x}"))
        ax.set_xlabel(str(x)); ax.set_ylabel(str(y))
        fig.text(
            0.01, 0.01,
            _chart_scope(
                len(plotted), chart_exactness,
                str(plan.get("scope") or "comparable rows"),
            ),
            fontsize=8, color="#52606d",
        )
        plt.show()
        records.append({
            "chart": f"relationship: {x} vs {y}",
            "plotted_rows": len(plotted), "exactness": chart_exactness,
        })

    if not records and not reasons:
        reasons.append(
            "Fields are too sparse or low-cardinality for an honest chart; use the table instead."
        )
    return records, reasons, _runtime_chart_manifest(records, reasons, prefix="adaptive")
````

Cardinality and mapping evidence show which entities and categories are actually represented.

````python

configured_coverage = READER.get("coverage")
source_entity_coverage = (
    (READER.get("source_aggregates") or {}).get("entity_coverage")
    if isinstance(READER.get("source_aggregates"), dict) else None
)
coverage_rows = []
if isinstance(source_entity_coverage, dict) and source_entity_coverage:
    coverage_rows.extend([
        {
            "column": column,
            "distinct": values.get("distinct_count"),
            "top_values": bounded(values.get("top_values"), 1600),
            "exactness": "exact over source_rows",
            "source_rows": SOURCE_ROWS,
        }
        for column, values in source_entity_coverage.items()
    ])

# Numeric codes are often valid entity axes but are intentionally absent from
# categorical vocabulary profiling.  Include only axes explicitly configured
# for entity display, and compute them from the attached frozen frame.
configured_entity_columns = []
configured_entity_columns.extend(as_list(READER.get("preview_strata")))
for kpi in as_list(READER.get("dashboard_kpis")):
    if isinstance(kpi, dict) and str(kpi.get("kind") or "") == "distinct_entities":
        configured_entity_columns.extend(as_list(kpi.get("columns")))
chart_plan = READER.get("chart_plan") or {}
for chart in as_list(chart_plan.get("charts") if isinstance(chart_plan, dict) else chart_plan):
    if isinstance(chart, dict):
        configured_entity_columns.extend(as_list(chart.get("series_dimensions")))
existing_columns = {str(row.get("column")) for row in coverage_rows}
if FRAME is not None:
    for column in dict.fromkeys(map(str, configured_entity_columns)):
        if column in existing_columns or column not in FRAME:
            continue
        series = FRAME[column]
        lower = column.lower()
        entity_like = (
            lower.endswith(("_code", "_id"))
            or any(token in lower for token in (
                "commodity", "country", "entity", "geography", "market",
                "port", "region", "state", "station", "territory",
            ))
        )
        if not entity_like:
            continue
        counts = series.dropna().value_counts().head(10)
        coverage_rows.append({
            "column": column,
            "distinct": int(series.nunique(dropna=True)),
            "top_values": bounded(
                [{"value": str(value), "rows": int(count)} for value, count in counts.items()],
                1600,
            ),
            "exactness": (
                "exact over source_rows"
                if EXACTNESS == "exact" and ANALYSIS_ROWS == SOURCE_ROWS
                else f"{EXACTNESS} over analyzed_rows"
            ),
            "source_rows": SOURCE_ROWS,
        })

vocabularies = metric_value("entity_vocabulary_coverage", "vocabularies", {}) or {}
for column, values in sorted(vocabularies.items()):
    if str(column) in {str(row.get("column")) for row in coverage_rows}:
        continue
    coverage_rows.append(
        {
            "column": column,
            "distinct": values.get("distinct_count"),
            "top_values": bounded(values.get("top_values"), 1600),
            "rare_categories": values.get("rare_category_count"),
            "rare_row_rate": values.get("rare_row_rate"),
            "normalization_collisions": len(values.get("normalisation_collisions", {})),
            "exactness": EXACTNESS,
        }
    )
if coverage_rows:
    display(pd.DataFrame(coverage_rows).sort_values("distinct", ascending=False).head(30))
elif isinstance(configured_coverage, list) and configured_coverage:
    display(pd.DataFrame(configured_coverage))
elif isinstance(configured_coverage, dict) and configured_coverage:
    display(pd.DataFrame([{"measure": key, "value": bounded(value, 1000)} for key, value in configured_coverage.items()]))
else:
    display(
        pd.DataFrame([{"status": "not assessed", "reason": "No vocabulary profile is available."}])
    )
interpretation(
    "Vocabulary checks support governed mappings; matching strings alone do not authorize a join."
)
````

## 5. Missingness and validity

Nulls, sentinels, infinities, zeros, and negatives are separated so structural gaps are not mistaken for signal.

````python

configured_missingness = READER.get("missingness")
source_column_statistics = (
    (READER.get("source_aggregates") or {}).get("column_statistics")
    if isinstance(READER.get("source_aggregates"), dict) else None
)
if isinstance(source_column_statistics, dict) and source_column_statistics:
    missing_frame = pd.DataFrame([
        {
            "column": column,
            "null_count": values.get("null_count"),
            "null_rate": values.get("null_rate"),
            "denominator": SOURCE_ROWS,
            "exactness": "exact over source_rows",
        }
        for column, values in source_column_statistics.items()
    ]).sort_values(["null_rate", "column"], ascending=[False, True])
elif isinstance(configured_missingness, list) and configured_missingness:
    missing_frame = pd.DataFrame(configured_missingness)
elif FRAME is not None and ANALYSIS_ROWS > 0:
    missing_frame = pd.DataFrame({
        "column": FRAME.columns,
        "null_count": [int(FRAME[column].isna().sum()) for column in FRAME.columns],
        "null_rate": [float(FRAME[column].isna().mean()) for column in FRAME.columns],
        "denominator": ANALYSIS_ROWS,
        "exactness": EXACTNESS,
    }).sort_values(["null_rate", "column"], ascending=[False, True])
else:
    missing_frame = pd.DataFrame([
        {"status": "not assessed", "reason": "No positive row denominator; rates are not shown as zero."}
    ])
missing_display = missing_frame.copy()
if "null_rate" in missing_display:
    missing_display["null_rate"] = missing_display["null_rate"].map(rate_display)
display(missing_display)
validity = None if OUTPUT_QUARANTINE else metric_value(
    "missingness_validity", "validity_summary", None
)
if OUTPUT_QUARANTINE:
    display(Markdown(
        "**Validity evidence:** numeric output-value validity is quarantined. "
        "The null/coverage QA above remains in scope and does not inspect output magnitudes."
    ))
elif validity:
    display(Markdown("**Validity evidence.** " + bounded(validity, 5000)))
````

## 9. PIT and leakage: what was knowable when?

Observation time, release time, knowledge time, and final/latest values are kept distinct.

````python

configured_pit = first_value(
    READER.get("pit_summary"), READER.get("pit"), READER.get("pit_notes")
)
if isinstance(configured_pit, list) and configured_pit:
    display(pd.DataFrame(configured_pit))
elif isinstance(configured_pit, dict) and configured_pit:
    display(pd.DataFrame([{"question": key, "answer": bounded(value, 1400)} for key, value in configured_pit.items()]))
else:
    display(pd.DataFrame([
        {"question": "Knowledge column", "answer": CONTRACT.get("knowledge_date_col") or "not governed"},
        {"question": "Knowledge semantics", "answer": CONTRACT.get("knowledge_semantics") or "not governed"},
        {"question": "Publication lag", "answer": CONTRACT.get("publication_lag_days") if CONTRACT.get("publication_lag_days") is not None else "not governed"},
        {"question": "Vintage retention", "answer": CONTRACT.get("vintage_retention") or "not declared"},
        {"question": "Source-level feature gate", "answer": "eligible" if metric_value("pit_leakage", "feature_eligible_source", False) else "blocked / diagnostic only"},
    ]))
hazards = text_items(OVERLAY.get("known_hazards"))
if hazards:
    display(Markdown("### Leakage and semantic hazards\n\n" + "\n".join(f"- {item}" for item in hazards)))
if OUTPUT_QUARANTINE and FRAME is not None and not FRAME.empty:
    reference_dates = []
    for item in as_list(PROVENANCE.get("object_sample")):
        if isinstance(item, dict) and item.get("last_modified"):
            parsed = pd.to_datetime(item["last_modified"], errors="coerce", utc=True)
            if not pd.isna(parsed):
                reference_dates.append(parsed)
    campaign_match = re.search(r"((?:19|20)\d{6})", str(PROVENANCE.get("campaign_id") or ""))
    if campaign_match:
        parsed = pd.to_datetime(campaign_match.group(1), format="%Y%m%d", errors="coerce", utc=True)
        if not pd.isna(parsed):
            reference_dates.append(parsed)
    governed_dates = []
    for column in ("as_of_date", "prediction_as_of_date", "prediction_date"):
        if column not in FRAME or column in VALUE_COLUMNS:
            continue
        parsed = pd.to_datetime(FRAME[column], errors="coerce", utc=True)
        if parsed.notna().any():
            governed_dates.append((column, parsed.max()))
    if reference_dates and governed_dates:
        reference = max(reference_dates)
        future_dates = [(column, value) for column, value in governed_dates if value > reference]
        if future_dates:
            column, value = max(future_dates, key=lambda item: item[1])
            display(Markdown(
                "**Output coverage warning.** "
                f"`{column}` reaches `{value.date()}`, after the latest reproducible source-object/"
                f"campaign capture reference `{reference.date()}`. This may be an intentional "
                "forecast horizon, but it must not be read as historical availability. "
                "No prediction or outcome values were inspected."
            ))
interpretation(
    "A useful source is not automatically PIT-safe. Prototype features only after their cutoff-time availability is explicit."
)
````

## 11. Feature opportunities and anti-features

Proposals are review records, not enabled features. Timing and lag rules are shown next to each idea.

````python

candidates = SUMMARY.get("feature_candidates", [])
if OUTPUT_QUARANTINE:
    assert not candidates and PROFILE.get("disposition") == "excluded_leakage"
candidate_frame = pd.DataFrame(candidates)
candidate_columns = [
    "candidate_id", "classification", "mechanism", "source_columns",
    "aggregation_window", "lag", "unit", "knowledge_time_rule", "readiness",
]
if not candidate_frame.empty:
    # Prefer breadth across source measures.  Candidate generation commonly
    # emits several transforms per measure; a reader preview of the first six
    # would otherwise hide every later feature-bearing column.
    representative_indexes = []
    seen_surfaces = set()
    for index, candidate in candidate_frame.iterrows():
        source_columns = as_list(candidate.get("source_columns"))
        transformation = candidate.get("transformation")
        primary = (
            transformation.get("column")
            if isinstance(transformation, dict) and transformation.get("column")
            else (source_columns[0] if source_columns else candidate.get("candidate_id"))
        )
        surface = str(primary)
        if surface in seen_surfaces:
            continue
        seen_surfaces.add(surface)
        representative_indexes.append(index)
        if len(representative_indexes) == 6:
            break
    visible_candidates = candidate_frame.loc[representative_indexes, [
        column for column in candidate_columns if column in candidate_frame
    ]]
    display(visible_candidates)
    if len(candidate_frame) > len(visible_candidates):
        display(Markdown(
            f"*Showing {len(visible_candidates)} representative ideas out of "
            f"{len(candidate_frame)} review-only candidates. The complete catalog is in "
            "`feature_candidates.yaml`.*"
        ))
else:
    rationale = first_value(
        READER.get("no_candidate_rationale"), OVERLAY.get("no_candidate_reason"),
        (SUMMARY.get("feature_opportunity_map", {}).get("no_candidate_rationale", {}) or {}).get("value"),
        default="No evidence-backed candidate survived source-quality gates.",
    )
    display(pd.DataFrame([{"candidate_status": "none", "reason": rationale}]))

anti_features = text_items(READER.get("anti_features"))
anti_summary = READER.get("anti_feature_summary")
if isinstance(anti_summary, dict):
    for entry in anti_summary.get("entries", []):
        if isinstance(entry, dict) and entry.get("statement"):
            text = str(entry["statement"])
            if entry.get("reason"):
                text += " - " + str(entry["reason"])
            anti_features.append(text)
else:
    anti_features += text_items(anti_summary)
anti_features += text_items(
    (SUMMARY.get("feature_opportunity_map", {}).get("do_not_derive", {}) or {}).get("value")
)
if OUTPUT_QUARANTINE:
    anti_features.append("Do not derive features from model predictions or output-plane values.")
display(Markdown(
    "### Do not derive\n\n" + (
        "\n".join(f"- {item}" for item in dict.fromkeys(anti_features))
        if anti_features else "- No table-specific anti-feature was supplied; PIT and semantic gates still apply."
    )
))
display(Markdown(
    "**Promotion boundary:** a separate human review is required; this EDA never edits production feature configuration."
))
````
